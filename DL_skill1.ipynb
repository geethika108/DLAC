```python
from ultralytics import YOLO
from PIL import Image
import matplotlib.pyplot as plt
import os
import shutil
```

```python
!wget -q -O traffic.jpg https://ultralytics.com/images/bus.jpg

print("Traffic image downloaded!")

img = Image.open("traffic.jpg")
plt.figure(figsize=(12, 7))
plt.imshow(img)
plt.axis("off")
plt.show()
```

```python
model = YOLO("yolo11n.pt")

print("VisionPort AI Model Loaded Successfully!")
```

```python
results = model.predict(
    source="traffic.jpg",
    conf=0.25,
    save=False
)

print("AI Traffic Detection Completed!")
```

```python
result_image = results[0].plot()

plt.figure(figsize=(14, 8))
plt.imshow(result_image[:, :, ::-1])
plt.title("VisionPort - AI Traffic Detection")
plt.axis("off")
plt.show()
```

```python
names = model.names

for box in results[0].boxes:
    class_id = int(box.cls[0])
    confidence = float(box.conf[0])

    print(
        f"Detected: {names[class_id]} | "
        f"Confidence: {confidence:.2%}"
    )
```

```python
os.makedirs("VisionPort", exist_ok=True)

output_path = "VisionPort/traffic_ai_detected.jpg"

Image.fromarray(result_image[:, :, ::-1]).save(output_path)

print("Detected image saved successfully!")
```
