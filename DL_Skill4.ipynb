```python
import torch

device = "cuda" if torch.cuda.is_available() else "cpu"

a = torch.tensor([10, 20, 30], device=device)
b = torch.tensor([1, 2, 3], device=device)

result = a * b

print("Device:", device)
print("Tensor A:", a)
print("Tensor B:", b)
print("Multiplication Result:", result)
```

```python
import torch

x = torch.tensor(5.0, requires_grad=True)

y = x ** 2

y.backward()

print("Input value:", x.item())
print("Output value:", y.item())
print("Gradient:", x.grad.item())
```

```python
import torch
import platform
import json
import os

environment_report = {
    "python_version": platform.python_version(),
    "pytorch_version": torch.__version__,
    "cuda_available": torch.cuda.is_available(),
    "gpu_name": torch.cuda.get_device_name(0) if torch.cuda.is_available() else "Not Available",
    "operating_system": platform.system(),
    "device_used": "cuda" if torch.cuda.is_available() else "cpu"
}

report_path = "/content/VisionPort_Skill4/reports/environment_report.json"

with open(report_path, "w") as file:
    json.dump(environment_report, file, indent=4)

print(json.dumps(environment_report, indent=4))
print("\nEnvironment report saved successfully!")
```
